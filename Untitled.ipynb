### Introduction
In search of the best neighborhood in Pittsburgh, we decided to focus on three metrics. The metrics were Safety, Affordability, and Recreation. We split up these metrics and found datasets for firearm seizures, playgrounds, and affordability score.


### Best Neighborhood
In conclusion, we found that Upper Lawrenceville was the best all around neighborhood based on all three of our metrics, being 23rd in firearm seizures, 11th in playgrounds, and 13th in affordability score.
