```python
import matplotlib.pyplot as plt
import pandas as pd

#loads the data set
df = pd.read_csv('Firearmseizures.csv')

print('Total Seizure Incidents:', len(df))
print('Total Neighborhoods Recorded:', df['neighborhood'].nunique())
print('-' * 30)

#Metric: few seizures == good neighborhood
# Count how many seizures occurred in each neighborhood
counts = df['neighborhood'].value_counts()
# Sort neighborhoods from safest (fewest seizures) to highest seizures
sorted_counts = counts.sort_values()

# Create a DataFrame and assign points from 90 down to 3
total_neighborhoods = len(sorted_counts)
points_df = sorted_counts.reset_index()
points_df.columns = ['Neighborhood', 'Seizure_Count']
points_df['Points'] = range(90, 90 - total_neighborhoods, -1)

# Print ALL 88 neighborhoods with their points
print(points_df.to_string(index=False))

# Split into the 10 safest (lowest) and 10 highest-seizure neighborhoods
safest_10 = counts.tail(10)
highest_10 = counts.head(10)


print('\n TOP 10 SAFEST NEIGHBORHOODS (FEWEST SEIZURES) ')
print(safest_10)

print('\n 10 HIGHEST-SEIZURE NEIGHBORHOODS ')
print(highest_10)

# Plot 1: Top 10 Safest Neighborhoods
plt.figure(figsize=(8, 5))
safest_10.plot(kind='barh', color='green')
plt.title('Top 10 Safest Pittsburgh Neighborhoods (Fewest Firearm Seizures)')
plt.xlabel('Number of Seizure Incidents')
plt.ylabel('Neighborhood')
plt.tight_layout()
plt.show()

# Plot 2: 10 Highest-Seizure Neighborhoods (For Contrast)
plt.figure(figsize=(8, 5))
highest_10.plot(kind='barh', color='red')
plt.title('10 Pittsburgh Neighborhoods with Most Firearm Seizures')
plt.xlabel('Number of Seizure Incidents')
plt.ylabel('Neighborhood')
plt.tight_layout()
plt.show()
```
